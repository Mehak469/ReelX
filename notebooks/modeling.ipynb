### Converte Tags Into Vectors and then find simlairty

```python
import sys
sys.path.append('../src')

import pandas as pd
import numpy as np
from preprocessor import stem #From src/prproceesing.py
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
movies=pd.read_csv("../data/processed/Processed_Movies.csv")
movies.head()
```

```python
movies.info()
```

### Apply stemming/lemmatizaion(fatsre speed required and just need to find similarity so stemming is better)
If we do Vectorization directly then From this you  can see that there are words that are like action,actions,actress,actresses,actor,actors,adventure,adventures,adventurous etc. So we need to do stemming or lemmatization to reduce the number of unique words. We will do that in the next step.(Before this vectorization we can also do stemming or lemmatization to reduce the number of unique words.)

```python
movies['tags']=movies['tags'].apply(stem)
movies['tags'][0]
```

```python
cv=CountVectorizer(max_features=5000, stop_words='english')
```

```python
bow=cv.fit_transform(movies['tags']).toarray()
bow.shape
```

```python
print("Unique words in the corpus:",cv.get_feature_names_out())
print("Number of unique words in the corpus:",len(cv.vocabulary_))
```

```python
df=pd.DataFrame(bow,columns=cv.get_feature_names_out())
df.to_csv("../data/processed/BOW.csv",index=False) 
# From this i can see that there are words that are like action,actions,actress,actresses,actor,actors,adventure,adventures,adventurous etc. So we need to do stemming or lemmatization to reduce the number of unique words. We will do that in the next step.(Before this vectorization we can also do stemming or lemmatization to reduce the number of unique words.))
```

### Now,Calculate the distance between Vectors
**Approaches**
1. Cosine distance(1-Cos_sim) -->distance (directly proportional) 1/cos_sim
2. Euclidean Distance(Tip to tip ditance -> Not recommended Cuz working in a hih dimenion(5000 Dimensional Data here))

```python
similarity=cosine_similarity(bow)
# shape(4806,4806)
```

```python
similarity[2]
```

```python
def recommend(movie):
    index=movies[movies['title']==movie].index[0]

    distances=similarity[index]

    movies_list=sorted(list(enumerate(distances)), reverse=True,key=lambda x:x[1])[1:6]
    
    for i in movies_list:
        print(movies.iloc[i[0]].title)
 
```

```python
recommend("Batman Begins")
```
