### Converte Tags Into Vectors and then find simlairty

```python
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer
```

```python
movies=pd.read_csv("../data/processed/Processed_Movies.csv")
movies.head()
```

```python
movies.info()
```

```python
cv=CountVectorizer()
```

```python
bow=cv.fit_transform(movies['tags'])
bow.shape
```

```python
print("Unique words in the corpus:",cv.vocabulary_)
print("Number of unique words in the corpus:",len(cv.vocabulary_))
```
